```python
import numpy as np
```

```python
def f(x):
    """Computes f(x)=x^3-x^2-1 given x"""
    return x**3 - x**2 - 1

def df(x):
    """Computes df/dx(x)=3x^2-2x"""
    return 3*(x**2)-2*x
```

```python
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """Finds roots to given function f and initial value x0 with Newton Iteration"""
    if abs(f(x0)) < epsilon:
        print(f"Found root in 0 iterations")
        print(x0)
        return 
    xn=x0
    n=0
    while (abs(f(xn)) >= epsilon) & (n < 30):
        if df(xn) == 0:
            print("Uh oh, division by 0 error")
            print('Iteration failed')
            return None
        xn = xn - f(xn)/df(xn)
        n+=1
    if n < 30:
        print(f"Found root in {n} iterations")
        print(xn)
        return 
    else:
        print('Iteration failed')
        return None
```

```python
# Part 3
print("Test division by zero case.")
newton(f, df, 0)

print("Epsilon is 1e-6:")
newton(f, df, 6)
print("Epsilon is 1e-8:")
newton(f, df, 6, 1e-8)

print("Epsilon is 1e-6:")
newton(f, df, 300)
print("Epsilon is 1e-8:")
newton(f, df, 300, 1e-8)

print("Epsilon is 1e-6:")
newton(f, df, 7000)
print("Epsilon is 1e-8:")
newton(f, df, 7000, 1e-8)
```

We see that the further away the initial value is from the zero, the more iterations it take for Newton Iteration to find the zero. Smaller the epsilon, more iterations it take to find the zero, but it also returns a more accurate zero.
